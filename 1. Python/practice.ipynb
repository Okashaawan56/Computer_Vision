List variable

```python
fruits=['apple', 'banana', 'grapes']
print (fruits)
print(fruits[0])
print(fruits[1])
print(fruits[-1])
print(fruits[-2])
print(fruits[0:3:2])
expensses=[10,20,30,40,50]
sale=[20,40,60,80,100]
profit=[]
for i in range(len(sale)):
    prof=sale[i]-expensses[i]
    profit.append(prof)
print(sale)
print(expensses)
print(profit)
# F stream printing
for i in range(len(sale)):
    print(f"sale={sale[i]} expensses={expensses[i]} and profit={profit[i]}")
```

Tuples

```python
# it is same as list but in list we can do changings but the tuple is immutable
color_a=(123,0,123)
color_b=(12,100,103)
color_c=(200,20,155)
mixed=[]
for i in range(len(color_a)):
    mix=(color_a[i]+color_b[i]+color_c[i])//3
    mixed.append(mix)
print(mixed)
mixed=tuple(mixed)
print(mixed)
```

```python
def calculator(number1,number2,operator):
    if operator=='+':
        print(number1+number2)
    elif operator=='-':
        print(number1-number2)
    elif operator=='*':
        print(number1-number2)
    else:
        print('wrong operator')
a=int(input('enter first number'))
b=int(input('enter second number'))
o=input('enter operator')
calculator(a,b,o)
```
